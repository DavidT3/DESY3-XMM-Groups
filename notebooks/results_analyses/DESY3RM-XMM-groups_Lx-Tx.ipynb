## Examining DESY3 group $L_{\rm{X}}$ and $T_{\rm{X}}$ results

### Cleaning invalid entries

<span style='color:red'>**This should probably all be done prior to getting here, in the measurement notebook - actually highlights an issue with the XGA-LTR, as no way should entries with negative Tx errors be considered valid - possibly happened on the last measurement after R500 had converged?**</span>

```python
desy3rm_groups = desy3rm_groups[(desy3rm_groups['Tx_500+'] > 0) & (desy3rm_groups['Tx_500-'] > 0) & 
                                (desy3rm_groups['Lx52_500+'] > 0) & (desy3rm_groups['Lx52_500-'] > 0)]
# Oh dear there is a real big Lx error, again this won't be here in the end
desy3rm_groups = desy3rm_groups[(desy3rm_groups['Lx52_500+'] < 1e+47) & (desy3rm_groups['Lx52_500-'] < 1e+47)]
desy3rm_groups
```

### Defining a 'high quality' group sample

```python
good_txerr_msk = (((desy3rm_groups['Tx_500-'] / desy3rm_groups['Tx_500']) < 0.25) & 
                  ((desy3rm_groups['Tx_500+'] / desy3rm_groups['Tx_500']) < 0.25))

# Doesn't make much sense, but some entries have very well constrained temperatures and crappy luminosities, so
#  for now I am making another mask to do the same percentage error selection as with temperatures
good_lxerr_msk = (((desy3rm_groups['Lx52_500-'] / desy3rm_groups['Lx52_500']) < 0.25) & 
                  ((desy3rm_groups['Lx52_500+'] / desy3rm_groups['Lx52_500']) < 0.25))

desy3rm_groups_txerr25_lxerr25 = desy3rm_groups[good_txerr_msk & good_lxerr_msk].reset_index(drop=True)
desy3rm_groups_txerr25_lxerr25
```

## Investigating measurements of $L_{\rm{X}}$-$T_{\rm{X}}$ that may belong in the cluster regime

Some of the data points in the figures shown above are firmly in the 'cluster regime' of temperatures and luminosities, clearly intermingling with the high-$T_{\rm{X}}$/$L_{\rm{X}}$ end of the SDSS sample we are comparing to. We attempt to separate some of those data points out in order to investigate those 'groups' in more detail.

Identifying those data points will also be useful for $\lambda_{\rm{RM}}$ (optical richness) relations that will be built, as we know that each of those data points has a $\lambda_{\rm{RM}}\leq20$ (as that is how we selected the sample), and that will result in some data points with very low richnesses compared to their high temperatures and luminosities.

```python
# Eyeballed from Figure 10 of Paul's paper - NOT TAKING INTO ACCOUNT THE E(z)^-1 scaling
tx_500_rich20_lim = Quantity(2.3, 'keV')
lx52_500_rich20_lim = Quantity(0.2e+44, 'erg/s')
```

### Uncertainties $<25\%$ subsample

```python
cl_reg_msk = ((desy3rm_groups_txerr25_lxerr25['Lx500ce_0.5-2.0'] >= lx52_500_rich20_lim.value) & 
              (desy3rm_groups_txerr25_lxerr25['Tx_500'] >= tx_500_rich20_lim.value))
print(cl_reg_msk.sum())
print(str(round((cl_reg_msk.sum() / cl_reg_msk.shape[0]*100), 1)) + '%')
```

#### Visualization

```python
plt.figure(figsize=(9, 9))
plt.minorticks_on()
plt.tick_params(which='both', direction='in', top=True, right=True)

desy3_grp_lx500_52 = (Quantity(desy3rm_groups_txerr25_lxerr25[['Lx52_500', 'Lx52_500-','Lx52_500+']].values, 'erg/s')*
                          (desy3rm_groups_txerr25_lxerr25['E'].values[..., None]**-1))
desy3_grp_tx500 = Quantity(desy3rm_groups_txerr25_lxerr25[['Tx_500', 'Tx_500-', 'Tx_500+']].values, 'keV')

plt.errorbar(desy3_grp_tx500[:, 0].value, desy3_grp_lx500_52[:, 0].value, yerr=desy3_grp_lx500_52[:, 1:].value.T, 
             xerr=desy3_grp_tx500[:, 1:].value.T, capsize=2, fmt='x', color='steelblue', label='DESY3RM Groups')

# Bodge to plot the high-lx/tx points slightly differently
plt.errorbar(desy3_grp_tx500[cl_reg_msk, 0].value, desy3_grp_lx500_52[cl_reg_msk, 0].value, 
             yerr=desy3_grp_lx500_52[cl_reg_msk, 1:].value.T, 
             xerr=desy3_grp_tx500[cl_reg_msk, 1:].value.T, capsize=2, fmt='D', color='teal', 
             label=r'DESY3RM Groups [High $L_{\rm{X}}$ & $T_{\rm{X}}$]')

# AND NOW THE SDSSRM-XCS SAMPLE AS A COMPARISON 
sdss_clust_lx500_52 = (Quantity(sdssrm_xcs_clust[['Lx52_500', 'Lx52_500-','Lx52_500+']].values, 'erg/s')*
                       (sdssrm_xcs_clust['E'].values[..., None]**-1))*1e+44
sdss_clust_tx500 = Quantity(sdssrm_xcs_clust[['Tx_500', 'Tx_500-', 'Tx_500+']].values, 'keV')

plt.errorbar(sdss_clust_tx500[:, 0].value, sdss_clust_lx500_52[:, 0].value, yerr=sdss_clust_lx500_52[:, 1:].value.T, 
             xerr=sdss_clust_tx500[:, 1:].value.T, capsize=2, fmt='d', color='darkgoldenrod', label='SDSSRM-XCS Clusters', 
             alpha=0.7)

plt.ylabel(r"$E(z)^{-1}L^{0.5-2.0}_{\rm{X,500}}$", fontsize=24)
plt.xlabel(r"$T_{\rm{X,500}}$", fontsize=24)

plt.xscale("log")
plt.yscale("log")

plt.gca().xaxis.set_major_formatter(FuncFormatter(lambda inp, _: '{:g}'.format(inp)))
plt.gca().xaxis.set_minor_formatter(FuncFormatter(lambda inp, _: '{:g}'.format(inp)))

plt.legend(fontsize=20)
plt.tight_layout()
plt.savefig(gen_fig_out_path + "desy3-groups_sdssrm-clusters_lx-tx_lt25percerr-cluster-regime-LxTx-split.pdf")
plt.show()
```
